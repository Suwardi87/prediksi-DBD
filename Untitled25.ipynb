```python
import openpyxl
import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

pd.set_option('display.width', 160)
pd.set_option('display.max_columns', 20)

wb = openpyxl.load_workbook(FILE_NAME, data_only=True)

LABEL_MAP = {'Rendah': 0, 'Sedang': 1, 'Tinggi': 2}
INV_LABEL = {0: 'Rendah', 1: 'Sedang', 2: 'Tinggi'}
BINARY_FEATURES = {'jk'}

def entropy(*probs):
    t = 0.0
    for p in probs:
        if p > 0:
            t -= p * math.log2(p)
    return t

def class_entropy(counts):
    total = sum(counts)
    if total == 0:
        return 0.0
    return entropy(*[c/total for c in counts])

FEATURE_ALIAS = {
    'usia': 'usia', 'Usia': 'usia',
    'lama_rawat': 'lama_rawat', 'Lama Rawat Inap': 'lama_rawat',
    'jk': 'jk', 'Jenis Kelamin': 'jk',
    'jumlah_kasus': 'jumlah_kasus', 'Jumlah Kasus Per Bulan': 'jumlah_kasus',
    'tingkat_resiko': 'tingkat_resiko', 'Tingkat Resiko': 'tingkat_resiko',
}

SHEET_NAMES = []
for i in range(1, 16):
    name = f'Pohon {i}' if i != 8 else 'Pohon 8 '
    SHEET_NAMES.append(name)

pohon_features = {}
for sn in SHEET_NAMES:
    ws = wb[sn]
    headers = [ws.cell(1, c).value for c in range(1, ws.max_column + 1)]
    nos = [i for i, h in enumerate(headers) if h == 'No']
    if len(nos) < 2:
        continue
    rs = nos[1]
    feats = []
    for c in range(rs + 1, ws.max_column):
        h = headers[c]
        if h and h in FEATURE_ALIAS and FEATURE_ALIAS[h] != 'tingkat_resiko':
            feats.append(FEATURE_ALIAS[h])
        elif h == 'Perhitungan root':
            break
    pohon_features[sn] = feats

print(f"Sheet names used: {SHEET_NAMES}")
print(f"Total: {len(SHEET_NAMES)} sheets")
for sn in SHEET_NAMES:
    print(f"  {sn:>9s} | {wb[sn].max_row or 0} baris | fitur: {pohon_features.get(sn, [])}")

def baca_data_pohon(sheet_name):
    ws = wb[sheet_name]
    headers = [ws.cell(1, c).value for c in range(1, ws.max_column + 1)]
    nos = [i for i, h in enumerate(headers) if h == 'No']
    if len(nos) < 2:
        return pd.DataFrame()
    rs = nos[1]
    col_map = {}
    for c in range(rs, ws.max_column):
        h = headers[c]
        if h and h in FEATURE_ALIAS:
            col_map[c] = FEATURE_ALIAS[h]
        elif h == 'Perhitungan root':
            break
    data = []
    for r in range(2, ws.max_row + 1):
        no_val = ws.cell(r, rs + 1).value
        if no_val is None or not isinstance(no_val, (int, float)):
            break
        row = {}
        for col_idx, feat_name in col_map.items():
            raw = ws.cell(r, col_idx + 1).value
            if feat_name in ('usia', 'lama_rawat', 'jumlah_kasus'):
                if raw is not None:
                    s = str(raw).strip().lower().replace('hari', '').strip()
                    try:
                        row[feat_name] = float(s)
                    except:
                        row[feat_name] = 0.0
                else:
                    row[feat_name] = 0.0
            elif feat_name == 'jk':
                try:
                    row[feat_name] = int(float(raw))
                except:
                    row[feat_name] = 0
            elif feat_name == 'tingkat_resiko':
                row[feat_name] = str(raw).strip() if raw else ''
        if 'tingkat_resiko' in row and row['tingkat_resiko'] in LABEL_MAP:
            data.append(row)
    if not data:
        return pd.DataFrame()
    df = pd.DataFrame(data)
    df['target_enc'] = df['tingkat_resiko'].map(LABEL_MAP)
    return df

def find_best_3way_split(df, feature):
    values = sorted(df[feature].unique())
    if len(values) <= 1:
        return None, None, 0, 0, {}
    
    n = len(df)
    target = 'target_enc'
    
    root_counts = [int((df[target] == c).sum()) for c in range(3)]
    root_entropy = class_entropy(root_counts)
    
    if feature in BINARY_FEATURES:
        t1 = t2 = 0.5
        left = df[df[feature] < t1]
        mid = df[(df[feature] >= t1) & (df[feature] <= t2)]
        right = df[df[feature] > t2]
        left_e = class_entropy([int((left[target] == c).sum()) for c in range(3)]) if len(left) > 0 else 0
        mid_e = class_entropy([int((mid[target] == c).sum()) for c in range(3)]) if len(mid) > 0 else 0
        right_e = class_entropy([int((right[target] == c).sum()) for c in range(3)]) if len(right) > 0 else 0
        weighted = (len(left)/n)*left_e + (len(mid)/n)*mid_e + (len(right)/n)*right_e
        gain = root_entropy - weighted
        detail = {
            'left_n': len(left), 'mid_n': len(mid), 'right_n': len(right),
            'left_ent': round(left_e, 6), 'mid_ent': round(mid_e, 6), 'right_ent': round(right_e, 6),
            'weighted_ent': round(weighted, 6)
        }
        return t1, t2, round(gain, 6), round(root_entropy, 6), detail
    
    midpoints = [(values[i] + values[i+1]) / 2.0 for i in range(len(values) - 1)]
    best_gain = -1
    best_t1 = best_t2 = None
    best_detail = {}
    
    for i in range(len(midpoints)):
        for j in range(i + 1, len(midpoints)):
            t1, t2 = midpoints[i], midpoints[j]
            left = df[df[feature] < t1]
            mid = df[(df[feature] >= t1) & (df[feature] <= t2)]
            right = df[df[feature] > t2]
            if len(left) == 0 or len(right) == 0:
                continue
            left_e = class_entropy([int((left[target] == c).sum()) for c in range(3)]) if len(left) > 0 else 0
            mid_e = class_entropy([int((mid[target] == c).sum()) for c in range(3)]) if len(mid) > 0 else 0
            right_e = class_entropy([int((right[target] == c).sum()) for c in range(3)]) if len(right) > 0 else 0
            weighted = (len(left)/n)*left_e + (len(mid)/n)*mid_e + (len(right)/n)*right_e
            gain = root_entropy - weighted
            if gain > best_gain:
                best_gain = gain
                best_t1, best_t2 = t1, t2
                best_detail = {
                    'left_n': len(left), 'mid_n': len(mid), 'right_n': len(right),
                    'left_ent': round(left_e, 6), 'mid_ent': round(mid_e, 6), 'right_ent': round(right_e, 6),
                    'weighted_ent': round(weighted, 6)
                }
    return best_t1, best_t2, round(best_gain, 6), round(root_entropy, 6), best_detail

def perhitungan_tree(df, feature_list):
    if df.empty or not feature_list:
        return pd.DataFrame()
    results = []
    for feat in feature_list:
        t1, t2, gain, root_ent, detail = find_best_3way_split(df, feat)
        if t1 is not None:
            results.append({
                'fitur': feat,
                't1': t1, 't2': t2,
                'IG': gain,
                'root_entropy': root_ent,
                'weighted_entropy': detail.get('weighted_ent', 0),
                'left_n': detail.get('left_n', 0),
                'mid_n': detail.get('mid_n', 0),
                'right_n': detail.get('right_n', 0),
            })
    if not results:
        return pd.DataFrame()
    res_df = pd.DataFrame(results).sort_values('IG', ascending=False).reset_index(drop=True)
    return res_df
```

```python
feat_display = {'usia': 'Usia', 'lama_rawat': 'Lama\nRawat', 'jk': 'JK', 'jumlah_kasus': 'Jumlah\nKasus'}

def gambar_stump_prop(ax, judul, fitur, t1, t2, left_df, mid_df, right_df):
    ax.set_xlim(0, 10); ax.set_ylim(0, 7); ax.axis('off')
    ax.set_title(judul, fontsize=8.5, fontweight='bold', pad=2)

    total = len(left_df) + len(mid_df) + len(right_df)
    if total == 0: return
    p_left = len(left_df)/total; p_mid = len(mid_df)/total; p_right = len(right_df)/total
    
    total_w = 8.0
    lw = max(1.5, total_w * p_left)
    mw = max(1.5, total_w * p_mid)
    rw = max(1.5, total_w * p_right)
    s = lw + mw + rw
    lw, mw, rw = lw/s*total_w, mw/s*total_w, rw/s*total_w
    
    root_w = 5; root_x = (10-root_w)/2
    root_box = mpatches.FancyBboxPatch((root_x, 5.2), root_w, 1.0, boxstyle="round,pad=0.08", fc='#5B6E8A', ec='#333')
    ax.add_patch(root_box)
    display_name = feat_display.get(fitur, fitur)
    thr_text = f"< {t1:.1f}" if t1==t2 else f"< {t1:.1f} | {t1:.1f}-{t2:.1f} | > {t2:.1f}"
    ax.text(5, 5.7, f"{display_name}\n{thr_text}", ha='center', va='center', color='white', fontsize=7.5, fontweight='bold')

    gap = 0.3
    left_x = gap + lw/2
    mid_x = left_x + lw/2 + gap + mw/2
    right_x = mid_x + mw/2 + gap + rw/2
    total_span = rw + 2*gap + mw + gap + lw
    offset = (10 - total_span) / 2
    left_x += offset; mid_x += offset; right_x += offset
    left_x0 = left_x - lw/2; mid_x0 = mid_x - mw/2; right_x0 = right_x - rw/2

    ax.plot([5, left_x], [5.2, 2.8], color='#555', lw=1.2)
    ax.plot([5, mid_x], [5.2, 2.8], color='#555', lw=1.2)
    ax.plot([5, right_x], [5.2, 2.8], color='#555', lw=1.2)
    ax.text(left_x, 4.2, f'{len(left_df)} ({p_left*100:.0f}%)', ha='center', fontsize=6.5, color='#555', fontweight='bold')
    ax.text(mid_x, 4.2, f'{len(mid_df)} ({p_mid*100:.0f}%)', ha='center', fontsize=6.5, color='#555', fontweight='bold')
    ax.text(right_x, 4.2, f'{len(right_df)} ({p_right*100:.0f}%)', ha='center', fontsize=6.5, color='#555', fontweight='bold')

    def make_pie(sub_df, x0, y0, w, h):
        if len(sub_df) == 0:
            box = mpatches.FancyBboxPatch((x0, y0), w, h, boxstyle="round,pad=0.05", fc='#ddd', ec='#999')
            ax.add_patch(box)
            ax.text(x0+w/2, y0+h/2, '0', ha='center', va='center', fontsize=8, color='#999')
            return
        counts = [int((sub_df['target_enc'] == c).sum()) for c in range(3)]
        total_c = sum(counts)
        if total_c == 0: return
        cum = 0
        for ci, (cl, color) in enumerate(zip(['Rendah','Sedang','Tinggi'], ['#2d8a4e','#e9a820','#c0392b'])):
            cnt = counts[ci]
            if cnt == 0: continue
            frac = cnt / total_c
            bw = max(0.5, w * frac)
            box = mpatches.FancyBboxPatch((x0 + cum, y0), bw, h, boxstyle="round,pad=0.02", fc=color, ec='white', lw=0.5)
            ax.add_patch(box)
            if bw > 1.0:
                ax.text(x0 + cum + bw/2, y0 + h/2, f'{cnt}', ha='center', va='center', fontsize=7, color='white', fontweight='bold')
            cum += bw

    bar_h = 1.2; bar_y = 0.8
    make_pie(left_df, left_x0, bar_y, lw, bar_h)
    make_pie(mid_df, mid_x0, bar_y, mw, bar_h)
    make_pie(right_df, right_x0, bar_y, rw, bar_h)
    
    labels = ['Rendah','Sedang','Tinggi']
    colors = ['#2d8a4e','#e9a820','#c0392b']
    leg_y = 0.15
    for li, (ll, lc) in enumerate(zip(labels, colors)):
        ax.add_patch(mpatches.FancyBboxPatch((0.3+li*1.8, leg_y), 0.4, 0.3, boxstyle="round,pad=0.02", fc=lc, ec='none'))
        ax.text(0.85+li*1.8, leg_y+0.15, ll, fontsize=6, va='center', color='#555')
```

```python
fig, axes = plt.subplots(3, 5, figsize=(24, 13))
axes = axes.flatten()

for i, sn in enumerate(SHEET_NAMES):
    df = baca_data_pohon(sn)
    if df.empty: continue
    feat = pohon_features.get(sn, [])
    if not feat: continue
    hasil_df = perhitungan_tree(df, feat)
    if hasil_df.empty: continue
    root = hasil_df.iloc[0]
    f = root['fitur']; t1, t2 = root['t1'], root['t2']
    left_df = df[df[f] < t1]
    mid_df = df[(df[f] >= t1) & (df[f] <= t2)]
    right_df = df[df[f] > t2]
    gambar_stump_prop(axes[i], f'Pohon {i+1} (IG={root["IG"]:.4f})', f, t1, t2, left_df, mid_df, right_df)

for j in range(i+1, 15): axes[j].axis('off')
plt.tight_layout()
plt.show()
```

## Penentuan Pohon Terbaik
