```python
import openpyxl, pandas as pd, numpy as np, math
import matplotlib.pyplot as plt, matplotlib.patches as mpatches
pd.set_option('display.width', 160); pd.set_option('display.max_columns', 20)

wb = openpyxl.load_workbook(FILE_NAME, data_only=True)
SHEET_NAMES = [f'Pohon {i}' for i in range(1, 16)]

# --- Baca langsung dari Excel (konsisten dengan aplikasi web, tanpa hardcode) ---
def _find_hdr_row(ws, key):
    for r in range(1, 250):
        for c in range(14, 22):
            v = ws.cell(r, c).value
            if v and isinstance(v, str) and v.strip().upper() == key:
                return r
    return None

def baca_threshold(sheet_name):
    ws = wb[sheet_name]
    hdr = _find_hdr_row(ws, 'JUMLAH KASUS PER BULAN')
    if hdr is None:
        return None, None
    t1 = ws.cell(hdr + 1, 16).value
    t2 = ws.cell(hdr + 1, 17).value
    t1 = float(t1) if isinstance(t1, (int, float)) else None
    t2 = float(t2) if isinstance(t2, (int, float)) else None
    return t1, t2

def baca_gain(sheet_name):
    ws = wb[sheet_name]
    for r in range(150, min(ws.max_row + 1, 250)):
        for c in range(15, 18):
            v = ws.cell(r, c).value
            if v and isinstance(v, str) and v.strip().upper() == 'IG':
                ig = ws.cell(r + 1, c).value
                ea = ws.cell(r + 1, c + 1).value
                if isinstance(ig, (int, float)) and isinstance(ea, (int, float)):
                    return float(ig), float(ea)
    return None, None

BAB4_POHON_CONFIG = {}
BAB4_GAIN = {}
BAB4_ENTROPY = {}
for n in range(1, 16):
    t1, t2 = baca_threshold(f'Pohon {n}')
    ig, ea = baca_gain(f'Pohon {n}')
    BAB4_POHON_CONFIG[n] = {'feature': 'jumlah_kasus', 't1': t1, 't2': t2}
    BAB4_GAIN[n] = ig
    BAB4_ENTROPY[n] = ea

# Evaluasi MAE/RMSE/R2 dari sheet PenentuanPohonterbaik
BAB4_EVAL = {}
ws_pp = wb['PenentuanPohonterbaik']
for r in range(2, ws_pp.max_row + 1):
    no = ws_pp.cell(r, 1).value
    if not isinstance(no, (int, float)):
        continue
    mae = ws_pp.cell(r, 3).value
    rmse = ws_pp.cell(r, 4).value
    r2 = ws_pp.cell(r, 5).value
    if isinstance(mae, (int, float)) and isinstance(rmse, (int, float)) and isinstance(r2, (int, float)):
        BAB4_EVAL[int(no)] = {'MAE': float(mae), 'RMSE': float(rmse), 'R2': float(r2)}

# Data Uji + prediksi pohon terbaik (dari BAB IV)
BAB4_DATA_UJI_JK = [12, 12, 7, 7, 18, 18, 18, 18, 21, 3]
BAB4_DATA_UJI_ACTUAL = [2, 2, 1, 1, 3, 3, 3, 3, 3, 1]
BAB4_BEST_PRED = [2, 2, 1, 1, 2, 2, 2, 2, 2, 1]
BEST_TREE_NUM = 6

LABEL_NAMES = {1: 'Rendah', 2: 'Sedang', 3: 'Tinggi'}

# Format 4 desimal dipotong (truncate), konsisten dengan aplikasi web
def trunc4(v):
    if v is None: return None
    return math.trunc(v * 10000) / 10000

def fmt4(v):
    if v is None: return '-'
    return f'{trunc4(v):.4f}'

def class_entropy(counts):
    total = sum(counts)
    if total == 0: return 0.0
    return sum(-(c/total)*math.log2(c/total) for c in counts if c > 0)

def baca_data_pohon(sheet_name):
    ws = wb[sheet_name]
    data = []
    for r in range(2, 200):
        no = ws.cell(r, 8).value
        if no is None or not isinstance(no, (int, float)): break
        jk = ws.cell(r, 11).value
        target = ws.cell(r, 13).value
        if jk is not None and target is not None:
            data.append({'jumlah_kasus': float(jk), 'tingkat_resiko': int(target)})
    return pd.DataFrame(data)

def compute_split(df, t1, t2):
    n = len(df)
    if n == 0: return None
    root_counts = [0, 0, 0]
    for t in df['tingkat_resiko'].tolist():
        root_counts[int(t) - 1] += 1
    root_ent = class_entropy(root_counts)
    left  = df[df['jumlah_kasus'] < t1]
    mid   = df[(df['jumlah_kasus'] >= t1) & (df['jumlah_kasus'] <= t2)]
    right = df[df['jumlah_kasus'] > t2]
    lc = [int((left['tingkat_resiko']==c+1).sum()) for c in range(3)]
    mc = [int((mid['tingkat_resiko']==c+1).sum()) for c in range(3)]
    rc = [int((right['tingkat_resiko']==c+1).sum()) for c in range(3)]
    le = class_entropy(lc); me = class_entropy(mc); re = class_entropy(rc)
    weighted = (len(left)/n)*le + (len(mid)/n)*me + (len(right)/n)*re
    gain = root_ent - weighted
    return {'t1':t1,'t2':t2,'IG':gain,'root_entropy':root_ent,
            'weighted_entropy':weighted,'root_counts':root_counts,
            'left_n':len(left),'mid_n':len(mid),'right_n':len(right),
            'left_counts':lc,'mid_counts':mc,'right_counts':rc}

print('Setup OK — Sheets:', [s for s in wb.sheetnames if s.startswith('Pohon')])
print('Threshold Pohon 6:', fmt4(BAB4_POHON_CONFIG[6]['t1']), '/', fmt4(BAB4_POHON_CONFIG[6]['t2']))
```

```python
# Tabel 4.18 — Hasil Tabel Perhitungan (semua pohon: Jumlah Kasus Per Bulan)
rows = []
for i, sn in enumerate(SHEET_NAMES):
    num = i + 1
    df = baca_data_pohon(sn)
    cfg = BAB4_POHON_CONFIG[num]
    result = compute_split(df, cfg['t1'], cfg['t2'])
    if result:
        rows.append({
            'No': num, 'Sampel': f'Sampel {num}',
            'Fitur': 'Jumlah Kasus Per Bulan',
            't1': cfg['t1'], 't2': cfg['t2'],
            'Gain_Hitung': result['IG'],
            'Entropy_Hitung': result['weighted_entropy'],
            'Gain_BAB4': BAB4_GAIN[num],
            'Entropy_BAB4': BAB4_ENTROPY[num],
        })
tabel418 = pd.DataFrame(rows)
print('=== Tabel 4.18 Hasil Tabel Perhitungan ===')
print(tabel418[['No','Sampel','t1','t2','Gain_Hitung','Gain_BAB4','Entropy_Hitung','Entropy_BAB4']].to_string(index=False))
```

```python
# Tabel 4.35 — Evaluasi MAE/RMSE/R² per pohon (dari BAB IV)
eval_rows = []
for num in range(1, 16):
    e = BAB4_EVAL[num]
    eval_rows.append({'No': num, 'Sampel': f'Sampel {num}',
                      'MAE': e['MAE'], 'RMSE': e['RMSE'], 'R2': e['R2']})
tabel435 = pd.DataFrame(eval_rows)
print('=== Tabel 4.35 Evaluasi per Pohon (MAE/RMSE/R²) ===')
print(tabel435.to_string(index=False))
print()
best_r2 = tabel435.loc[tabel435['R2'].idxmax()]
print(f"Pohon Terbaik: Sampel {int(best_r2['No'])} (R² = {best_r2['R2']})")
```

```python
def gambar_stump_3way(ax, judul, t1, t2, left_counts, mid_counts, right_counts):
    ax.set_xlim(0, 10); ax.set_ylim(0, 9.0); ax.axis('off')
    ax.set_title(judul, fontsize=9, fontweight='bold', pad=6)
    total = sum(left_counts) + sum(mid_counts) + sum(right_counts)
    cls_names = ['Rendah', 'Sedang', 'Tinggi']

    # Root (ungu) — judul + total & threshold, seperti aplikasi web
    root_w, root_h = 6.8, 1.7
    rx0 = (10 - root_w) / 2; ry0 = 7.1
    rcx = rx0 + root_w / 2
    ax.add_patch(mpatches.FancyBboxPatch((rx0, ry0), root_w, root_h,
        boxstyle='round,pad=0.08', fc='#7667B2', ec='white', lw=2))
    ax.text(rcx, ry0 + root_h*0.66, 'Jumlah Kasus Per Bulan', ha='center', va='center',
            color='white', fontsize=9, fontweight='bold')
    ax.text(rcx, ry0 + root_h*0.28, f'Total: {total} data \u2014 Threshold: {fmt4(t1)} / {fmt4(t2)}',
            ha='center', va='center', color='#DDD0F0', fontsize=6.5)

    # Panah dari root ke 3 cabang (dengan head, seperti aplikasi web)
    leaf_w, leaf_h = 3.1, 3.2
    ly0 = 0.6
    centers = [
        (1.8, left_counts,  '#C87B48', f'< {fmt4(t1)}', 'Kiri'),
        (5.0, mid_counts,   '#5B9E6D', f'{fmt4(t1)} - {fmt4(t2)}', 'Tengah'),
        (8.2, right_counts, '#4B74A5', f'> {fmt4(t2)}', 'Kanan'),
    ]
    for cx, counts, color, thr_txt, branch_name in centers:
        total_b = sum(counts)
        maj = max(range(3), key=lambda i: counts[i]) if total_b > 0 else None
        ax.annotate('', xy=(cx, ly0 + leaf_h), xytext=(rcx, ry0 + root_h*0.18),
                    arrowprops=dict(arrowstyle='-|>', color='#94a3b8', lw=1.5))
        bx0 = cx - leaf_w/2
        ax.add_patch(mpatches.FancyBboxPatch((bx0, ly0), leaf_w, leaf_h,
            boxstyle='round,pad=0.08', fc=color, ec='white', lw=1.8))
        # nama cabang (atas kotak, uppercase)
        ax.text(cx, ly0 + leaf_h*0.86, branch_name.upper(), ha='center', va='center',
                color='white', fontsize=6.5, fontweight='bold', alpha=0.9)
        # threshold (baris ke-2)
        ax.text(cx, ly0 + leaf_h*0.62, thr_txt, ha='center', va='center',
                color='white', fontsize=7, fontweight='bold')
        # jumlah kelas mayoritas (baris ke-3) - seperti PDF (16/55/30)
        mv = counts[maj] if (maj is not None and total_b > 0) else 0
        ax.text(cx, ly0 + leaf_h*0.40, f'{cls_names[maj]} : {mv}', ha='center', va='center',
                color='white', fontsize=7, fontweight='bold')
```

```python
print('=' * 64)
for num in range(1, 16):
    df = baca_data_pohon(f'Pohon {num}')
    cfg = BAB4_POHON_CONFIG[num]
    res = compute_split(df, cfg['t1'], cfg['t2'])
    rc = res['root_counts']
    print(f'Detail Pohon {num}')
    print(f'  Entropy Root            : {fmt4(res["root_entropy"])}')
    print(f'  Information Gain        : {fmt4(res["IG"])}')
    print(f'  Threshold               : t1={fmt4(cfg["t1"])}, t2={fmt4(cfg["t2"])}')
    print(f'  Distribusi Kelas (Root) : Rendah={rc[0]}, Sedang={rc[1]}, Tinggi={rc[2]}')
print('=' * 64)

fig, axes = plt.subplots(3, 5, figsize=(24, 14))
axes = axes.flatten()
for i, sn in enumerate(SHEET_NAMES):
    num = i + 1
    df = baca_data_pohon(sn)
    cfg = BAB4_POHON_CONFIG[num]
    result = compute_split(df, cfg['t1'], cfg['t2'])
    if result:
        gambar_stump_3way(axes[i], f'Pohon {num} (Gain={fmt4(BAB4_GAIN[num])})',
                          cfg['t1'], cfg['t2'],
                          result['left_counts'], result['mid_counts'], result['right_counts'])
plt.tight_layout()
plt.show()
```

## Pengujian Data Uji — Confusion Matrix, Accuracy, Precision, Recall, F1-Score

```python
y_actual = BAB4_DATA_UJI_ACTUAL
y_pred   = BAB4_BEST_PRED
print('=== Tabel 4.38 Data Uji (10 baris) ===')
for i in range(10):
    print(f"  {i+1:2d}. JK={BAB4_DATA_UJI_JK[i]:2d}  Aktual={LABEL_NAMES[y_actual[i]]:7s}  Prediksi={LABEL_NAMES[y_pred[i]]}")

print('\n=== Tabel 4.39 Hasil Kesalahan Diagnosa ===')
POSITIVE = 2
for i in range(10):
    a, p = y_actual[i], y_pred[i]
    if a == POSITIVE and p == POSITIVE: jenis = 'TP'
    elif a != POSITIVE and p == POSITIVE: jenis = 'FP'
    elif a != POSITIVE and p != POSITIVE: jenis = 'TN'
    else: jenis = 'FN'
    print(f"  {i+1:2d}. Aktual={a} ({LABEL_NAMES[a]:7s})  Prediksi={p} ({LABEL_NAMES[p]:7s})  {jenis}")

cm = [[0]*3 for _ in range(3)]
for a, p in zip(y_actual, y_pred):
    cm[a-1][p-1] += 1

print('\n=== Tabel 4.40 Confusion Matrix (Binary: Positif=Sedang) ===')
cm_binary = {}
for cls_name, cls_enc in [('Rendah',1), ('Sedang',2), ('Tinggi',3)]:
    positif = sum(1 for a,p in zip(y_actual,y_pred) if a==cls_enc and p==POSITIVE)
    negatif = sum(1 for a,p in zip(y_actual,y_pred) if a==cls_enc and p!=POSITIVE)
    cm_binary[cls_name] = {'positif': positif, 'negatif': negatif}
    print(f"  {cls_name:8s}: Positif={positif}  Negatif={negatif}  Total={positif+negatif}")

tp = sum(1 for a,p in zip(y_actual,y_pred) if a==POSITIVE and p==POSITIVE)
fp = sum(1 for a,p in zip(y_actual,y_pred) if a!=POSITIVE and p==POSITIVE)
tn = sum(1 for a,p in zip(y_actual,y_pred) if a!=POSITIVE and p!=POSITIVE)
fn = sum(1 for a,p in zip(y_actual,y_pred) if a==POSITIVE and p!=POSITIVE)
n = len(y_actual)

accuracy = (tp + tn) / n
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

print(f'\n=== Evaluasi Kinerja (Tabel 4.39-4.44) ===')
print(f'True Positive (TP)  : {tp}')
print(f'True Negative (TN)  : {tn}')
print(f'False Positive (FP) : {fp}')
print(f'False Negative (FN) : {fn}')
print(f'Accuracy  = (TP+TN)/Total = ({tp}+{tn})/{n} = {accuracy:.4f} = {accuracy*100:.0f}%')
print(f'Precision = TP/(TP+FP)    = {tp}/({tp}+{fp}) = {precision:.4f} = {precision*100:.2f}%')
print(f'Recall    = TP/(TP+FN)    = {tp}/({tp}+{fn}) = {recall:.4f} = {recall*100:.0f}%')
print(f'F1-Score  = 2xPxFx/(P+F)  = {f1:.4f} = {f1*100:.2f}%')
```
