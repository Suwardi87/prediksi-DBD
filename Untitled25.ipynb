```python
import openpyxl, pandas as pd, numpy as np, math
import matplotlib.pyplot as plt, matplotlib.patches as mpatches
pd.set_option('display.width', 160); pd.set_option('display.max_columns', 20)

wb = openpyxl.load_workbook(FILE_NAME, data_only=True)
SHEET_NAMES = [f'Pohon {i}' for i in range(1, 16)]

BAB4_POHON_CONFIG = {
    1:  {'feature': 'jumlah_kasus', 't1': 13.70, 't2': 29.60},
    2:  {'feature': 'jumlah_kasus', 't1': 13.06, 't2': 30.11},
    3:  {'feature': 'jumlah_kasus', 't1': 11.99, 't2': 25.70},
    4:  {'feature': 'jumlah_kasus', 't1': 13.05, 't2': 28.24},
    5:  {'feature': 'jumlah_kasus', 't1': 12.11, 't2': 26.17},
    6:  {'feature': 'jumlah_kasus', 't1': 11.39, 't2': 24.76},
    7:  {'feature': 'jumlah_kasus', 't1': 11.55, 't2': 25.22},
    8:  {'feature': 'jumlah_kasus', 't1': 11.31, 't2': 26.52},
    9:  {'feature': 'jumlah_kasus', 't1': 12.15, 't2': 26.26},
    10: {'feature': 'jumlah_kasus', 't1': 13.05, 't2': 27.48},
    11: {'feature': 'jumlah_kasus', 't1': 13.70, 't2': 26.60},
    12: {'feature': 'jumlah_kasus', 't1': 11.49, 't2': 24.24},
    13: {'feature': 'jumlah_kasus', 't1': 12.41, 't2': 24.75},
    14: {'feature': 'jumlah_kasus', 't1': 13.18, 't2': 27.21},
    15: {'feature': 'jumlah_kasus', 't1': 12.21, 't2': 25.26},
}
BAB4_GAIN = {1:0.5575,2:0.6771,3:0.6003,4:0.6208,5:0.5644,6:0.5482,7:0.5677,8:0.6409,9:0.5779,10:0.5858,11:0.5548,12:0.5720,13:0.4393,14:0.5903,15:0.4993}
BAB4_ENTROPY = {1:0.6324,2:0.5928,3:0.7622,4:0.5877,5:0.7614,6:0.8130,7:0.8084,8:0.7385,9:0.7528,10:0.7191,11:0.7247,12:0.8184,13:0.7731,14:0.7152,15:0.7839}
BAB4_EVAL = {
    1:{'MAE':19.33,'RMSE':21.648,'R2':-4.794}, 2:{'MAE':22.66,'RMSE':26.229,'R2':-11.286},
    3:{'MAE':23.66,'RMSE':31.984,'R2':-353.12}, 4:{'MAE':29,'RMSE':30.881,'R2':-137.44},
    5:{'MAE':18.66,'RMSE':21.556,'R2':-2.9305}, 6:{'MAE':17.66,'RMSE':22.708,'R2':-2.4378},
    7:{'MAE':28.66,'RMSE':36.72,'R2':-122.86}, 8:{'MAE':29,'RMSE':33.64,'R2':-12.99},
    9:{'MAE':29.66,'RMSE':36.33,'R2':-25.643},10:{'MAE':28,'RMSE':28.856,'R2':-95.077},
   11:{'MAE':11.333,'RMSE':13.114,'R2':-14.796},12:{'MAE':23.66,'RMSE':31.459,'R2':-233.39},
   13:{'MAE':20.33,'RMSE':22.575,'R2':-2.634},14:{'MAE':23,'RMSE':23.014,'R2':-11.161},
   15:{'MAE':18,'RMSE':23.122,'R2':-6.225},
}
BAB4_DATA_UJI_JK = [12, 12, 7, 7, 18, 18, 18, 18, 21, 3]
BAB4_DATA_UJI_ACTUAL = [2, 2, 1, 1, 3, 3, 3, 3, 3, 1]
BAB4_BEST_PRED = [2, 2, 1, 1, 2, 2, 2, 2, 2, 1]
BEST_TREE_NUM = 6

LABEL_NAMES = {1: 'Rendah', 2: 'Sedang', 3: 'Tinggi'}

def class_entropy(counts):
    total = sum(counts)
    if total == 0: return 0.0
    return sum(-(c/total)*math.log2(c/total) for c in counts if c > 0)

def baca_data_pohon(sheet_name):
    ws = wb[sheet_name]
    data = []
    for r in range(2, 200):
        no = ws.cell(r, 8).value
        if no is None or not isinstance(no, (int, float)): break
        jk = ws.cell(r, 11).value
        target = ws.cell(r, 13).value
        if jk is not None and target is not None:
            data.append({'jumlah_kasus': float(jk), 'tingkat_resiko': int(target)})
    return pd.DataFrame(data)

def compute_split(df, t1, t2):
    n = len(df)
    if n == 0: return None
    root_counts = [0, 0, 0]
    for t in df['tingkat_resiko'].tolist():
        root_counts[int(t) - 1] += 1
    root_ent = class_entropy(root_counts)
    left  = df[df['jumlah_kasus'] < t1]
    mid   = df[(df['jumlah_kasus'] >= t1) & (df['jumlah_kasus'] <= t2)]
    right = df[df['jumlah_kasus'] > t2]
    lc = [int((left['tingkat_resiko']==c+1).sum()) for c in range(3)]
    mc = [int((mid['tingkat_resiko']==c+1).sum()) for c in range(3)]
    rc = [int((right['tingkat_resiko']==c+1).sum()) for c in range(3)]
    le = class_entropy(lc); me = class_entropy(mc); re = class_entropy(rc)
    weighted = (len(left)/n)*le + (len(mid)/n)*me + (len(right)/n)*re
    gain = root_ent - weighted
    return {'t1':t1,'t2':t2,'IG':round(gain,4),'root_entropy':round(root_ent,4),
            'weighted_entropy':round(weighted,4),'left_n':len(left),'mid_n':len(mid),'right_n':len(right),
            'left_counts':lc,'mid_counts':mc,'right_counts':rc}

print('Setup OK — Sheets:', [s for s in wb.sheetnames if s.startswith('Pohon')])
```

```python
# Tabel 4.18 — Hasil Tabel Perhitungan (semua pohon: Jumlah Kasus Per Bulan)
rows = []
for i, sn in enumerate(SHEET_NAMES):
    num = i + 1
    df = baca_data_pohon(sn)
    cfg = BAB4_POHON_CONFIG[num]
    result = compute_split(df, cfg['t1'], cfg['t2'])
    if result:
        rows.append({
            'No': num, 'Sampel': f'Sampel {num}',
            'Fitur': 'Jumlah Kasus Per Bulan',
            't1': cfg['t1'], 't2': cfg['t2'],
            'Gain_Hitung': result['IG'],
            'Entropy_Hitung': result['weighted_entropy'],
            'Gain_BAB4': BAB4_GAIN[num],
            'Entropy_BAB4': BAB4_ENTROPY[num],
        })
tabel418 = pd.DataFrame(rows)
print('=== Tabel 4.18 Hasil Tabel Perhitungan ===')
print(tabel418[['No','Sampel','t1','t2','Gain_Hitung','Gain_BAB4','Entropy_Hitung','Entropy_BAB4']].to_string(index=False))
```

```python
def gambar_stump_3way(ax, judul, t1, t2, left_counts, mid_counts, right_counts):
    ax.set_xlim(0, 10); ax.set_ylim(0, 7); ax.axis('off')
    ax.set_title(judul, fontsize=7.5, fontweight='bold', pad=2)
    root_w, root_h = 5.5, 1.2
    rx0 = (10 - root_w) / 2; ry0 = 5.2
    rcx = rx0 + root_w / 2
    ax.add_patch(mpatches.FancyBboxPatch((rx0, ry0), root_w, root_h,
        boxstyle='round,pad=0.1', fc='#7667B2', ec='white', lw=1.5))
    ax.text(rcx, ry0 + root_h*0.65, 'Jumlah Kasus Per Bulan', ha='center', va='center',
            color='white', fontsize=6, fontweight='bold')
    ax.text(rcx, ry0 + root_h*0.28, f'< {t1:.1f}  |  {t1:.1f}-{t2:.1f}  |  > {t2:.1f}',
            ha='center', va='center', color='#DDD0F0', fontsize=5)
    leaf_w, leaf_h = 2.4, 2.5
    ly0 = 0.6
    centers = [(1.8, left_counts, '#C87B48', f'< {t1:.1f}', 'Kiri'),
              (5.0, mid_counts,  '#5B9E6D', f'{t1:.1f}-{t2:.1f}', 'Tengah'),
              (8.2, right_counts,'#4B74A5', f'> {t2:.1f}', 'Kanan')]
    cls_colors = ['#2d8a4e', '#e9a820', '#c0392b']
    cls_labels = ['Rendah', 'Sedang', 'Tinggi']
    for cx, counts, color, thr_txt, branch_label in centers:
        total = sum(counts)
        ax.plot([rcx, cx], [ry0, ly0 + leaf_h], color='#555', lw=1)
        ax.text((rcx + cx)/2, ry0 - 0.2, branch_label, ha='center', fontsize=5,
                color='#555', fontweight='bold',
                bbox=dict(boxstyle='round,pad=0.08', fc='white', ec='none', alpha=0.85))
        bx0 = cx - leaf_w/2
        ax.add_patch(mpatches.FancyBboxPatch((bx0, ly0), leaf_w, leaf_h,
            boxstyle='round,pad=0.1', fc=color, ec='white', lw=1))
        ax.text(cx, ly0 + leaf_h - 0.3, thr_txt, ha='center', va='center',
                color='white', fontsize=5.5, fontweight='bold')
        ax.text(cx, ly0 + leaf_h - 0.65, f'n = {total}', ha='center', va='center',
                color='white', fontsize=5)
        detail = f"Rendah:{counts[0]}  Sedang:{counts[1]}  Tinggi:{counts[2]}"
        ax.text(cx, ly0 + 0.6, detail, ha='center', va='center',
                color='white', fontsize=5.5, fontweight='bold')
        bar_y = ly0 + 0.2; bar_h = 0.18
        if total > 0:
            cum = bx0 + 0.15; avail = leaf_w - 0.3
            for ci in range(3):
                if counts[ci] == 0: continue
                seg = avail * counts[ci] / total
                ax.add_patch(mpatches.Rectangle((cum, bar_y), seg, bar_h,
                    fc=cls_colors[ci], ec='white', lw=0.3))
                cum += seg
```

```python
fig, axes = plt.subplots(3, 5, figsize=(24, 13))
axes = axes.flatten()
for i, sn in enumerate(SHEET_NAMES):
    num = i + 1
    df = baca_data_pohon(sn)
    cfg = BAB4_POHON_CONFIG[num]
    result = compute_split(df, cfg['t1'], cfg['t2'])
    if result:
        gambar_stump_3way(axes[i], f'Pohon {num} (Gain={BAB4_GAIN[num]:.4f})',
                          cfg['t1'], cfg['t2'],
                          result['left_counts'], result['mid_counts'], result['right_counts'])
plt.tight_layout()
plt.show()
```

## Pengujian Data Uji — Confusion Matrix, Accuracy, Precision, Recall, F1-Score

```python
y_actual = BAB4_DATA_UJI_ACTUAL
y_pred   = BAB4_BEST_PRED
print('=== Tabel 4.36 Data Uji (10 baris) ===')
for i in range(10):
    print(f"  {i+1:2d}. JK={BAB4_DATA_UJI_JK[i]:2d}  Aktual={LABEL_NAMES[y_actual[i]]:7s}  Prediksi={LABEL_NAMES[y_pred[i]]}")

print('\n=== Tabel 4.39 Hasil Kesalahan Diagnosa ===')
for i in range(10):
    a, p = y_actual[i], y_pred[i]
    if a == p: jenis = 'TP'
    elif a < p: jenis = 'FP'
    else: jenis = 'FN'
    print(f"  {i+1:2d}. Aktual={a} ({LABEL_NAMES[a]:7s})  Prediksi={p} ({LABEL_NAMES[p]:7s})  {jenis}")

cm = [[0]*3 for _ in range(3)]
for a, p in zip(y_actual, y_pred):
    cm[a-1][p-1] += 1

print('\n=== Tabel 4.40 Confusion Matrix ===')
print(f"{'':>12s} {'Pred=Rendah':>12s} {'Pred=Sedang':>12s} {'Pred=Tinggi':>12s} {'Total':>6s}")
labels = ['Rendah', 'Sedang', 'Tinggi']
for i in range(3):
    total_row = sum(cm[i])
    print(f"{'Act='+labels[i]:>12s} {cm[i][0]:>12d} {cm[i][1]:>12d} {cm[i][2]:>12d} {total_row:>6d}")

correct = sum(cm[c][c] for c in range(3))
total = len(y_actual)
tp = correct
fp = sum(cm[r][c] for r in range(3) for c in range(3) if r != c and cm[r][c] > 0 and c == max(range(3), key=lambda x: sum(cm[rr][x] for rr in range(3))))
fp_count = total - correct - sum(1 for i in range(10) if y_actual[i] != y_pred[i] and y_actual[i] > y_pred[i])
fn_count = sum(1 for i in range(10) if y_actual[i] != y_pred[i] and y_actual[i] > y_pred[i])
fp_count = sum(1 for i in range(10) if y_actual[i] != y_pred[i] and y_actual[i] < y_pred[i])

accuracy = correct / total
precision = tp / (tp + fp_count) if (tp + fp_count) > 0 else 0
recall = tp / (tp + fn_count) if (tp + fn_count) > 0 else 0
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

print(f'\n=== Evaluasi Kinerja (Tabel 4.40-4.44) ===')
print(f'True Positive (TP)  : {tp}')
print(f'False Positive (FP) : {fp_count}')
print(f'False Negative (FN) : {fn_count}')
print(f'Accuracy  = {tp}/{total} = {accuracy:.2f} = {accuracy*100:.0f}%')
print(f'Precision = {tp}/({tp}+{fp_count}) = {precision:.2f} = {precision*100:.0f}%')
print(f'Recall    = {tp}/({tp}+{fn_count}) = {recall:.2f} = {recall*100:.0f}%')
print(f'F1-Score  = 2x{precision:.2f}x{recall:.2f}/({precision:.2f}+{recall:.2f}) = {f1:.2f} = {f1*100:.0f}%')
```
